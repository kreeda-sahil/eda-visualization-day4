```python
# 1. Visualizing Outliers with Box Plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Box plot for Cost
sns.boxplot(x=df['approx_cost_for_two'], ax=axes[0], color='coral')
axes[0].set_title('Outliers: Approximate Cost for Two')
axes[0].set_xlabel('Cost (₹)')

# Box plot for Votes
sns.boxplot(x=df['votes'], ax=axes[1], color='purple')
axes[1].set_title('Outliers: Number of Votes')
axes[1].set_xlabel('Total Votes')

plt.tight_layout()
plt.show()


# 2. Quantifying Outliers using the IQR Method
def get_iqr_stats(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Count how many data points fall outside the bounds
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    outlier_count = len(outliers)
    percentage = (outlier_count / len(df)) * 100
    
    return upper_bound, outlier_count, percentage

cost_upper, cost_count, cost_pct = get_iqr_stats('approx_cost_for_two')
votes_upper, votes_count, votes_pct = get_iqr_stats('votes')

print(f"--- approx_cost_for_two ---")
print(f"Upper Bound for Normal Data: ₹{cost_upper}")
print(f"Number of Outliers: {cost_count} ({cost_pct:.2f}% of data)")

print(f"\n--- votes ---")
print(f"Upper Bound for Normal Data: {votes_upper} votes")
print(f"Number of Outliers: {votes_count} ({votes_pct:.2f}% of data)")
```
